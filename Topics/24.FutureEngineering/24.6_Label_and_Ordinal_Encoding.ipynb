#### Label Encoding
Label Encoding and oridinal encoding are two techniques used to encode categorical data as numerical data.

label encoding involves assigning a unique numeric label to each category in the variable.The labels are usually assigned in alphabetical order or based on the freqvency of the categories. for example, if we have categorical variable 'color' with three possible values (red,green,blue), we can represent as follows
1.Red:1
2.Green:2
3.Blue:3

```python
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
```

```python
# Create simple Df
df = pd.DataFrame({
    'color':['red','blue','green','green','red','blue']
})
```

```python
df
```

```python
lbl_encoder = LabelEncoder()
```

```python
lbl_encoder.fit_transform(df[['color']])
```

```python
# New Data
lbl_encoder.transform([['red']])
```

# DisAdvantage of Label:-
In above example we are assigned values, red=1,blue=2,green=3 so model may think 2>1 or 1<3 -> blue>red ...etc to overcomes this will go for ordinal Encoding (this wont happen normally but if there is a situation where we need assign ranks to perticular values to this categories we can defintly do using Ordinal Encoding)

### Ordinal Encoding:---
It is used to encode categorical data that have intrinsic order or ranking. In this technique, each category is assigned a numeric value based on its position in the order.For example, if we have a categorical value 'EducationLevel' with four possible values(high school,college,graduate,post-graduate) we can represent it using ordinal encoding as follows.

1.HighSchool:1
2.College:2
3.Graduate:3
4.PostGraduate:4

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.DataFrame({
    'size':['small','medium','large','medium','small','large']
})
```

```python
df
```

```python
encoder = OrdinalEncoder(categories=[['small','medium','large']])
```

```python
encoder.fit_transform(df[['size']])
```
