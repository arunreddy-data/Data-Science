#### Data Encoding

converting categories into numerical values for ML
1. Nominal/OHE Encoding
2. Label and Ordinary Encoding
3. Target Guided Ordinary Encoding

##### Nominal/OHE Encoding:-
One Hot Encoding, also known as nominal encoding, is technique used to represent categorical data as numerical data, which is more suitable for machine learning algorithms. In this technique each category represented as binary vector where the each bit corresponds to a unique category. for example, if we have a categorical variable color with three possible values(red,green,blue) we can represent one hot encoding as follows:

1.Red:[1,0,0]
2.Green:[0,1,0]
3.Blue:[0,0,1]

##### Disadvantages:-
When we more Categories this is not recomended because the size of matrix will get increases.
If matrix consists only 0,1 will call it as Spare Matrix it leads to overfitting (model gets trined well with data)

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
# Create simple Df
df = pd.DataFrame({
    'color':['red','blue','green','green','red','blue']
})
```

```python
df
```

```python
# Create Instance of OneHotEncoder
encoder = OneHotEncoder()
```

```python
# Perform fit and transform
encoded = encoder.fit_transform(df[['color']]).toarray()
```

```python
encoder_df =pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
encoder_df
```

```python
# For any new data
encoder.transform([['green']]).toarray()
```

```python
pd.concat([df,encoder_df],axis=1)
```
