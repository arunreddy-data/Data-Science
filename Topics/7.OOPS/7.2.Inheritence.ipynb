Inheritence is fundamental concept of OOP that allows a class to inherit attributes and methods from another class. This Lession covers single inheritence and Multiple inheritence, demonstrating how to create and use them in python.

```python
## Inheritnce
## Class
class Car:

    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype

    def drive(self):
        print(f"The Person will drive the {self.enginetype} car.")

class Tesla(Car):

    def __init__(self,windows,doors,enginetype,is_selfdriving):
        super().__init__(windows,doors,enginetype)
        self.is_selfdriving=is_selfdriving

    def selfdriving(self):
        print(f"Tesla Supports Self Driving : {self.is_selfdriving} ")


tesla1 =Tesla(4,5,"Electric",True)
tesla1.selfdriving()

car1 = Car(4,5,'Petrol')
car1.drive()

```

```python
# Multiple Inheritence
# When Class Inherits from more than one class.

class Animal:
    def __init__(self,name):
        self.name=name
    def speak():
        print("Sub Classes must implement this method.")

class Pet:
    def __init__(self,owner):
        self.owner= owner

# derived class
class Dog(Animal,Pet):
    def __init__(self,name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)

    def speak(self):
        return f"{self.name} say woof"

# Create an Object
dog = Dog('buddy','Arun')
print(dog.speak())
```
