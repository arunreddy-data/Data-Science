SQL and SQLite
SQL is standard language for managing and manupulating relational databases. SQLiteis a self-contained, serverless, and zero-configuration database engine that is widelyused for embedded database systems. In this lession, we will cover the basics of SQL and SQLite, including creating databases, tables, and performing various SQL operations.

```python
import sqlite3
print('Hell0o')
```

```python
# Connect to sqlite3 database
connection =sqlite3.connect('example.db')
connection
```

```python
cursor = connection.cursor()
cursor
```

```python
# Create table
cursor.execute(""" CREATE TABLE IF NOT EXISTS employees(
                    id integer Primary Key,
                    name vrachar(20) not null,
                    age integer,
                    department varchar(20))
               """)
# commit the chages
connection.commit()
```

```python
cursor.execute('''SELECT * FROM employees''')
```

```python
## Insert data into sqlite table.
cursor.execute('''
INSERT INTO employees(name,age,department)
VALUES ('Arun',22,'DataScientist')
''')
cursor.execute('''
INSERT INTO employees(name,age,department)
VALUES ('Kohli',32,'Sports')
''')
cursor.execute('''
INSERT INTO employees(name,age,department)
VALUES ('malli',28,'Tester')
''')
connection.commit()
```

```python
# Query the Data
cursor.execute("""SELECT * FROM employees""")
rows = cursor.fetchall()
print(rows)
for row in rows:
    print(row)
```

```python
# Update 
cursor.execute("""UPDATE employees SET name='Mallika' WHERE department='Tester'""")
connection.commit()
```
