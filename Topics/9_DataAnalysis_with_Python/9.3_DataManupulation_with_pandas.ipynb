Data Manipulation and Analysis with Pandas

Data Manipulation and Analysis are key tasks in any data science or data analysis project. Pandas provides a wide range of functions for data manipulation and analysis, making it easier to clean, transform and extract insights from data. In this lession, we will cover various data manipulation and analysis techniques using pandas.

```python
import pandas as pd

data_df = pd.read_csv('data.csv')
# fetch first 5 rows
data_df.head()
```

```python
data_df.describe()
```

```python
data_df.dtypes
```

```python
# Handling Missing Values.
data_df.isnull()
```

```python
data_df.isnull().any()
```

```python
data_df.isnull().sum()
```

```python
data_filled = data_df.fillna(0) # In place of null we are replacing with 0
```

```python
# filling missing values with mean of values
data_df ['Sales_fillNA'] = data_df['Sales'].fillna(data_df['Sales'].mean())
data_df
```

```python
# filling missing values with median
data_df['Value_fillNA'] = data_df['Value'].fillna(data_df['Value'].median())
data_df
```

```python
# Mode - it return Series even if we have single mode it will return Series
mode = data_df['Value'].mode()
mode # Series
```

```python
mode = data_df['Value'].mode().iloc[0]
data_df['Value_fillNA'] = data_df['Value'].fillna(mode)
data_df
```

```python
data_df[['Value', 'Value_fillNA']]
```

```python
# Renaming The columns
data_df=data_df.rename(columns={'Date':'Sales_Date'})
data_df.head()
```

```python
# 1. Does NOT change original
data_df.rename(columns={'Date': 'Sales_Date'})

# 2. Changes variable by assigning returned DataFrame
data_df = data_df.rename(columns={'Date': 'Sales_Date'})

# 3. Modifies original DataFrame
data_df.rename(columns={'Date': 'Sales_Date'}, inplace=True)
```

```python
# Changing DataType
# data_df['Value_new']=data_df['Value'].astype(int)  # If we have na values we can not chnage directly
data_df['Value_new'] =data_df['Value'].fillna(data_df['Value'].mean()).astype(int)
data_df.dtypes
```

```python
# Increase rate of product 
data_df['New_Value'] = data_df['Value'].apply(lambda x:x*2)
data_df.head()
```

```python
## Data Aggrigating and Grouping
grouped_sales = data_df.groupby('Product')['Value'].sum()
grouped_sales
```

```python
grouped_mean = data_df.groupby('Product')['Value'].mean()
grouped_mean
```

```python
grouped_sum = data_df.groupby(['Product','Region'])['Value'].sum()
grouped_sum
```

```python
## Aggrigate Multiple Functions
grouped_aggr = data_df.groupby('Region')['Value'].agg(['mean','sum','count'])
grouped_aggr
```

```python
groupe = data_df.groupby(['Product','Region'])[['Value','Sales']].sum()
groupe
```

```python
# Best way to do GROUPING

grouped = data_df.groupby(['Product','Region']).agg({'Value':'sum','Sales':'median'})
print(grouped)
```

```python
grouped = data_df.groupby(['Product']).agg({'Value':'max','Value':'min',
                                            'Value': lambda x:x.mode().iloc[0]})
# dict does't allow duplicate keys so will get only one result mode follow below approch
grouped
```

```python
grouped = data_df.groupby(['Product']).agg(
Max_Value = ('Value','max'),
Min_Value =('Value','min'),
mode_value =('Value', lambda x:x.mode().iloc[0])
)
grouped
```

```python
# Merging and Joining DF
df1 = pd.DataFrame({'key':['A','B','C'],'Value1':[1,2,3]})
df2 = pd.DataFrame({'key':['A','B','D'],'Value2':[4,5,6]})
df1
```

```python
df2
```

```python
# Merge DF on the key column
pd.merge(df1,df2)
```

```python
pd.merge(df1,df2,on='key',how='inner') # INNER JOIN 
```

```python
pd.merge(df1,df2,on='key',how='right')
```

```python
pd.merge(df1,df2,on='key',how='left') # LEFT JOIN
```

```python
pd.merge(df1,df2,on='key',how='outer')
```
