Pandas DataFrame and Series
Pandas is a powerful data manufulation library in python. widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame. A Series is one-dimensional array like object, while DataFrame is a two dimensional, size-mutable and potentially hetrogeneous tabular data structure with labeled axes(rows and columns).

```python
import pandas as pd
```

```python
# Series :- Pandas Series is one-dimensional array like object that can hold any datatype. 
# It is similar to a column in a table.

data =[1,2,3,4,5]
series= pd.Series(data)
print(series)
print(type(series))
```

```python
# Series from Dictionary (key's will becomes as index when we create series from dict)
data = {'a':1,'b':2,'c':3,'d':4}

series_dict = pd.Series(data)
print(series_dict)

```

```python
data =[10,20,30]
index = ['a','b','c']

pd.Series(data,index=index)
```

```python
# DataFrame
# Create a DataFrame from dict of list
data = {
    'Name':['Arun','Virat','Josh'],
    'Age': [22,38,34],
    'city':['Andhra','Delhi','Aus']
}
df = pd.DataFrame(data)
print(type(df))
print(df)
```

```python
# Converting DF to Numpy Array
import numpy as np

arr = np.array(df)
print(arr)
```

```python
# Create a Df from a List of Dictionary

data = [
    {'Name':'Arun','Age': 22,'city':'Andhra'},
    {'Name':'Virat','Age': 38,'city':'Delhi'},
    {'Name':'Josh','Age': 34,'city':'Aus'},
     {'Name':'salt','Age': 30,'city':'Eng'},
]
df = pd.DataFrame(data)
print(df)
```
