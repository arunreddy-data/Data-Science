Data Visualization With Seaborn

Seaborn is a python visulization library based on matplotlib that provides a high-level
interface for drawing attractive and informative statistical graphics. Seaborn helps in
creating complex visulization with just few lines of code. IN this lession will cover
basics of Seaborn, including creating various types of plots and customizing them.

```python
import seaborn as sns
```

```python
# Basic ploting with seaborn 
tips = sns.load_dataset('tips')
tips
```

```python
# Create Scatter plot 
import matplotlib.pyplot as plt

sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title("Scatter plot of taotalbill vs Tip")
```

```python
sns.lineplot(x='size',y='total_bill',data=tips)
plt.title("Line Plot of Total bill by size")
```

```python
# Categorical Variables

sns.barplot(x='day',y='total_bill',data=tips)
plt.title("Bar Plot of Total bill By Day")
plt.show()
```

```python
# Box Plot

sns.boxplot(x="day",y='total_bill',data=tips)
plt.show()
```

```python
# Violin Plot

sns.violinplot(x='day',y='total_bill',data=tips)
```

```python
# Histogram
sns.histplot(tips['total_bill'],bins=10,kde=True)  # to get curve plot
```

```python
# KDE
sns.kdeplot(tips['total_bill'],fill=True)
```

```python
# Pair Plot
sns.pairplot(tips)
```

```python
# Heat map  (corelation : relation b/w one feature to another feature.)
corr = tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
```

```python
import pandas as pd

sales_df =pd.read_csv('sales_data.csv')
sales_df.head()
```

```python
# plot total sales by product
plt.figure(figsize=(10,6))
# Here no need to do group by from pandas from seaplot we have estimator feature
sns.barplot(x='Product Category',y='Total Revenue',data=sales_df,estimator=sum)
plt.title("Total Sales By Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.show()
```

```python
# Total Sales by Region
plt.figure(figsize=(10,6))
# Here no need to do group by from pandas from seaplot we have estimator feature
sns.barplot(x='Region',y='Total Revenue',data=sales_df,estimator=sum)
plt.title("Total Sales By Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```
