PYTHON Logging:-
Logging is crucial aspect of any application, providing a way to track events, errors, and operartional information. Python's built-in logging module offers a flexible framework for emitting log messages from python programs. we will cover the basics of logging, including how to configure logging log levels, and best practices for using logging in Python Applications.

```python
import logging
# Configure the basic logging settings
logging.basicConfig(level=logging.DEBUG)

# Log messages
logging.debug("This is Debug message")
logging.info("This is info message")
logging.warning("This is a warning message.")
logging.error("This is Error Message")
logging.critical("This is critical message")
```

# LOG LEVELS:-
Python's logging module has several log levels indicating the severity of events, The default levels are:

-> DEBUG:- Detailed Information, typically of intrest only when diagnosing problems.
-> INFO :- Conformation that things are working as expected.
-> WARNING:- An indication that something unexcepted happened or indentative of some problem in near future (e.g. 'disk space is low'). The software is still working as expected.
-> ERROR:- Due to more serious problem, the software has not been able to perform some function.
-> CRITICAL:- A very serious Error, indicating that the program itself may be unable to continue running.

```python
# Configure Logging
import logging
logging.basicConfig(
    filename='app.log',
    filemode='w',
    level=logging.DEBUG,
                    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
                    datefmt='%Y-%m-%d %H:%M:%S')

logging.debug("This is Debug message")
logging.info("This is info message")
logging.warning("This is a warning message.")
logging.error("This is Error Message")
logging.critical("This is critical message")
```
