Abstraction

Abstraction is the concept of hiding the complex implementation and showing only the necessary features of an object. This helps in reducing programming complexity and effort.


```python
from abc import ABC,abstractmethod

## Abstract class
class Vehicle(ABC):
    def drive():
        print("The Vehicle is used for driving.")
    
    @abstractmethod
    def start_engine():
        pass

class Car(Vehicle):
    def start_engine(self):
        print("Car Engine Started.")

def operate_vehicle(vehicle):
    vehicle.start_engine()

car = Car()
operate_vehicle(car)
```
