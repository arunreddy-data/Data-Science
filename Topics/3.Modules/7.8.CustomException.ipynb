Custom Exception (raise and throw exception)

```python
class Error(Exception):
    pass

class dobException(Error):
    pass

year =int(input("Enter dob"))
age = 2026-year

try:   
    if age <=30 and age>20:
        print("Age is valid we can apply.")
    else:
        raise dobException
except dobException:
    print("Your not aligible (age should be 20-30)")
```
