Operator OverLoading
OperratorOverloading allows you to define the behavior of operators (+,-,*,/,..etc) for custom objects. you achieve this by overriding specific methods in your class.

```python
# Common Operator overloading Magic methods
'''
__add__(self,other): Add two objects using + Operator
__sub__(self,other): substract two objects using - Operator
__myl__(self,other): multiplies two objects using * Operator
__truediv__(self,other): divides two objects using / Operator
__eq__(self,other): check if two objects are equal using == Operator
__lt__(self,other): check if one object is less than another object using < Operator

'''
```

```python
## Mathamatical operations for Vector
class Vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __add__(self,other):
        return Vector(self.x+other.x,self.y+other.y)
    def __sub__(self,other):
            return Vector(self.x-other.x,self.y-other.y)
    def __mul__(self,other):
            return Vector(self.x*other.x,self.y*other.y)
    def __eq__(self,other):
            return self.x==other.x,self.y==other.y
    def __repr__(self):
        return f"Vector({self.x},{self.y})"

v1 = Vector(2,3)
v2 = Vector(4,5)
print(v1+v2)
print(v1-v2)
print(v1*v2)
print(v1==v2)
```
