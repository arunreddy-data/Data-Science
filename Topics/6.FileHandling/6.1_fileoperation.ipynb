File Operation- Read And Write Files

File Handling is a crucial of any programming language. Python provides built-in functions
and methods to read from and write to files, both text and binary.This lession will cover basics of file handling, including reading and writing text files and binary files.

```python
## Read a Whole File.
with open('example.txt','r') as file: ## r means read
    content = file.read()
    print(content)
```

```python
# Read File Line by Line
with open('example.txt','r') as file:
    for line in file:
        print(line.strip()) # strip will remove new line
```

```python
# Writing Of file (Over writing..)
with open('example.txt','w') as file:
   file.write('Hello World!\n')
   file.write('This is new Line') 
```

```python
# Writing a file (without Overwriting)
with open('example.txt','a') as file: # a - append 
    file.write("Append Operation taking Place..\n")
    file.write("Arun reddy")
    file.write("ViratKohli\n")
    file.write("Chase Master\n")
```

```python
# Writing List of lines to file
lines =['First line \n','Second line \n','Third Line \n']

with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
# Binary File.
# Writing to a binary file
data = b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file : # wb -> write binary
    file.write(data)
```

```python
# Binary File.
# reading a binary file
with open('example.bin','rb') as file : # wb -> write binary
    content = file.read()
    print(content)
```

```python
## Read the content from source text file and write to destination file.
## Coping a text file.

with open('example.txt','r') as file:
    content = file.read()
with open('destination.txt','w') as file:
    file.write(content)
```

The "w+" mode in python is used to open the file for both reading and writing. If the file does not exists, It will be created. If the File Exists, its content will be truncated (i.e. the file is overwritten.)

```python
## Writing and Reading the file

with open('example.txt','w+') as file:
    file.write("Hello World\n")
    file.write('This is New Line\n')

    ## Move the Cursor to the begining.
    file.seek(0)
    ## Read the content of the file.
    content = file.read()
    print(content)
```
